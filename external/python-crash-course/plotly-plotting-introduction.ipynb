# Plotly for small numerical investigations

This notebook uses deterministic synthetic values around the familiar root-finding target $f(x)=x^2-2$. They are examples for visualization, not a complete numerical-method implementation.

## Jupyter, Plotly Express, and interaction

In a notebook, assigning a figure to `fig` creates a Python object; displaying `fig` as the last expression or calling `fig.show()` asks the notebook renderer to display it. Plotly normally chooses a suitable interactive renderer for Jupyter environments. Start with `plotly.express` (`px`), the concise high-level interface. Each Express call returns a figure that can later receive careful customization; constructing raw `graph_objects` figures is an advanced route, not the beginner default.

Hover over points for values, drag to zoom, double-click inside a plot to reset, and use the modebar's home/reset control when present. Interactive details help inspection, but they do not replace numerical evidence.

```python
import numpy as np
import plotly.express as px

x = np.linspace(1.0, 2.0, 9)
y = x**2 - 2
fig = px.line(x=x, y=y, labels={"x": "x (unitless)", "y": "f(x) = x² − 2"}, title="A small Plotly figure")
fig.add_hline(y=0, line_dash="dash", line_color="black", annotation_text="zero reference")
fig.show()
```

**Text alternative and limitation.** The curve increases from negative to positive values and crosses the dashed zero reference near x = √2. The line makes a root candidate visible; it does not prove that a numerical method will reach the root or that a computed value is accurate.

---

## 1. Candidate points and a function curve

A scatter plot is useful for discrete candidate values. Here, a line for the known function provides context. Line segments on the curve show evaluated function values, while markers identify the candidate points; the visual difference does not depend on color alone.

```python
import numpy as np
import plotly.express as px

candidate_x = np.array([1.20, 1.35, 1.41, 1.414])
candidate_y = candidate_x**2 - 2
curve_x = np.linspace(1.0, 1.7, 160)
curve_y = curve_x**2 - 2

fig = px.scatter(
    x=candidate_x, y=candidate_y,
    labels={"x": "Candidate x (unitless)", "y": "Function value f(x)"},
    title="Candidate values against f(x) = x² − 2",
)
fig.update_traces(marker={"symbol": "diamond", "size": 11}, name="candidate points", showlegend=True)
fig.add_scatter(x=curve_x, y=curve_y, mode="lines", name="function curve", line={"color": "#1f77b4"})
fig.add_hline(y=0, line_dash="dash", line_color="black", annotation_text="f(x) = 0")
fig.show()
```

**Text alternative and limitation.** Diamond markers approach the dashed zero line as $x$ approaches about 1.414, while the blue curve crosses that line near the same location. This is a diagnostic comparison, not a fit test or proof that the candidates came from a correct or convergent algorithm.

---

## 2. Ordered iteration history

A line plot requires an ordered horizontal variable. Here, consecutive markers are iteration records, so the connecting segments indicate sequence—not values computed between iterations.

```python
import numpy as np
import plotly.express as px

iteration = np.arange(6)
x_n = np.array([1.000000, 1.500000, 1.416667, 1.414216, 1.414214, 1.414214])
fig = px.line(
    x=iteration, y=x_n, markers=True,
    labels={"x": "Iteration n", "y": "Recorded iterate xₙ (unitless)"},
    title="Ordered candidate history",
)
fig.update_traces(marker={"symbol": "circle", "size": 9}, line={"color": "#1f77b4"})
fig.add_hline(y=np.sqrt(2), line_dash="dash", line_color="black", annotation_text="reference √2")
fig.show()
```

**Text alternative and limitation.** The recorded iterates move from 1.0 to near the dashed √2 reference, with smaller visible changes late in the sequence. Small changes alone are not a stopping-rule justification and do not certify accuracy without a suitable error or residual analysis.

---

## 3. Error history on logarithmic axes

When positive errors span several scales, a logarithmic vertical axis can reveal their change. On a semilog plot, a nearly straight pattern can suggest geometric decay; on log-log axes, a nearly straight pattern suggests power-law scaling. These are different interpretations. Do not put signed or zero values on a logarithmic axis.

```python
import numpy as np
import plotly.express as px

iteration = np.arange(6)
absolute_error = np.array([4.142e-1, 8.579e-2, 2.453e-3, 2.124e-6, 1.595e-12, 1.0e-15])
fig = px.line(
    x=iteration, y=absolute_error, log_y=True, markers=True,
    labels={"x": "Iteration n", "y": "Absolute error |xₙ − √2| (unitless)"},
    title="Positive error history on a logarithmic vertical axis",
)
fig.update_traces(marker={"symbol": "square", "size": 9}, line={"color": "#d62728"})
fig.show()
```

**Text alternative and limitation.** The positive absolute error drops across many orders of magnitude as iteration increases. The logarithmic scale makes early and late changes visible together. The pattern is suggestive, but a plotted slope does not establish convergence order; rounding, asymptotic-regime assumptions, and a numerical-analysis argument still matter.

## 5. A small numerical grid heatmap

Heatmaps are appropriate when data naturally lie on a grid. The example maps the signed value of $x^2+y^2-1$ over a small square. A diverging palette is appropriate because zero is meaningful. Read coordinates and hover values as well as color: color is not the only encoding.

```python
import numpy as np
import plotly.express as px

grid = np.linspace(-1.2, 1.2, 9)
X, Y = np.meshgrid(grid, grid)
Z = X**2 + Y**2 - 1
fig = px.imshow(
    Z, x=grid, y=grid, origin="lower",
    color_continuous_scale="RdBu_r", color_continuous_midpoint=0,
    labels={"x": "x (unitless)", "y": "y (unitless)", "color": "x² + y² − 1"},
    title="Signed function values on a 9 × 9 grid",
    aspect="equal",
)
fig.show()
```

**Text alternative and limitation.** Grid values are negative near the origin, near zero around the unit circle, and positive near the corners; hovering reports the numeric value at each grid point. The colorbar and hover values support interpretation, but this coarse grid does not locate a contour accurately or establish an error bound.
