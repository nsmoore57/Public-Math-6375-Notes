# Dot Products and Matrix Operations

```python
import numpy as np
```

## Critical Operations:
- `np.dot()` vs `@` operator
- Difference between element-wise multiplication (`*`) and dot product

```python
v1 = np.array([1, 2])
v2 = np.array([3, 4])
print('vector 1:\n', v1)
print('vector 2:\n', v2)

# Dot product
print("Dot Product:", v1 @ v2) 
print("Another way to do the dot product:", v1.dot(v2)) 
print('Using the * operator (element-wise multiplication):\n', v1 * v2)
```

```python
# Matrix multiplication example
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print('matrix A:\n', A)
print('matrix B:\n', B)

print("Matrix Product:\n", A @ B)
print("Element-Wise Multiplication:\n", A * B)
```

```python
# Matrix-vector multiplication - notice the dimension and how it is displayed
print(A@v1)
```

## Transpose Operation
- Use `.T` attribute to get the transpose of a matrix.

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print("Original:\n", matrix)
print("Original shape:", matrix.shape)
print("Transposed:\n", matrix.T)
print("Transposed shape:\n", matrix.T.shape)
```

## Norm Calculations
### Vector and matrix norms:
- L1/L2 norms with `np.linalg.norm()`

```python
v = np.array([3, -4])
print('vector v:', v)
print("L2 norm (magnitude):", np.linalg.norm(v)) # Should be 5
print('L1 norm:', np.linalg.norm(v, ord=1)) # Should be 7

A = np.array([[1, 2], [3, 4]])
print('matrix A:\n', A)

print("1-Norm of matrix:", np.linalg.norm(A, ord=1))
print("infinity-Norm of matrix:", np.linalg.norm(A, ord=np.inf))
print("Frobenius norm of matrix:", np.linalg.norm(A, ord='fro'))
```
