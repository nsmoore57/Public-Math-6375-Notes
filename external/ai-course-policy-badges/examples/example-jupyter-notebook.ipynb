# Example Jupyter Notebook AI Use Badges

These examples show all three badge policies in standard Markdown cells.


![AI Not Permitted. Students must complete the marked work without AI assistance.](https://nsmoore57.github.io/ai-course-policy-badges/badges/ai-not-permitted.svg)

**AI Use: AI Not Permitted.**  
Complete this marked work without AI assistance. Use this for unaided understanding, foundational skills, or individual reasoning.


![AI Debugging Only. Students may use AI to understand errors, diagnose problems, and ask guiding questions, but not to generate a full solution.](https://nsmoore57.github.io/ai-course-policy-badges/badges/ai-debugging-only.svg)

**AI Use: AI Debugging Only.**  
You may use AI to understand error messages, diagnose problems, or ask guiding questions, but not to generate the solution.


![AI Permitted with Disclosure. Students may use AI as a tutor, assistant, brainstorming partner, or reviewer, but must disclose its use and understand submitted work.](https://nsmoore57.github.io/ai-course-policy-badges/badges/ai-permitted-disclosure.svg)

**AI Use: AI Permitted with Disclosure.**  
You may use AI as a tutor, coding assistant, brainstorming partner, or reviewer, but you must disclose its use and understand everything you submit.
